```python
import pandas as pd
project_id = "dh-global-sales-data-dev"
```

```python
query = """
select * from `dh-global-sales-data.instant_match_vertex.left_processed_PY_PE`
"""

df_py_pe = pd.read_gbq(query=query, project_id = project_id)
```

```python
for index, row in df_py_pe.iterrows():
    row_id = row["row_id"]
    try:
        _, contents = row_id.split("----")
    except ValueError as e:
        raise ValueError(f"row_id parsed incorrectly: {index}, {row_id}")
```
